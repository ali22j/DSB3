```python
import ultralytics
from ultralytics import YOLO

print("Ultralytics:", ultralytics.__version__)
model = YOLO("/content/best.pt")

print("Task:", model.task)
```

```python
results = model.train(
    data="/content/Solar-panels-seg-2/data.yaml",
    epochs=20,
    imgsz=640,
    batch=32,
    device=0,
    patience=10,
    optimizer="AdamW",
    lr0=0.001,
    lrf=0.01,
    warmup_epochs=1
)
```
